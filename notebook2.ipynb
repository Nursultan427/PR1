```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Iris.csv")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["sepal_width"], bins=15, edgecolor="black")

plt.title("Распределение ширины чашелистика")
plt.xlabel("Ширина чашелистика")
plt.ylabel("Частота")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.boxplot(df["sepal_width"])

plt.title("Boxplot ширины чашелистика")
plt.ylabel("Ширина чашелистика")

plt.show()
```

```python
df["sepal_width"].describe()
```

```python
species_counts = df["species"].value_counts()

plt.figure(figsize=(8, 5))

species_counts.plot(kind="bar", edgecolor="black")

plt.title("Количество цветков каждого вида")
plt.xlabel("Вид цветка")
plt.ylabel("Количество")

plt.xticks(rotation=0)
plt.show()
```


## Итоговая таблица: проблемы данных
